## 1.GPU 사용
- 파이토치 하면서 GPU사용하는 방법은 사용자가 고민안해도 되는건가?
- 파이토치는 사용자가 명시적으로 알려주지 않으면 기본적으로 모든 연산을 CPU에서 처리합니다.
- TensorFlow 같은 일부 프레임워크는 GPU가 있으면 자동으로 할당하기도 하지만, 파이토치는 "모든 제어권을 개발자에게 준다"는 철학을 가지고 있어서 데이터(Tensor)와 모델(Model)을 GPU 메모리로 직접 옮겨주어야 합니다.
- 파이토치에서 GPU를 쓰려면 보통 아래의 3단계 패턴을 무조건 작성하게 됩니다.

### 1). 1. GPU 장치 설정
- 먼저 현재 환경에서 GPU(CUDA)를 쓸 수 있는지 확인하고, 사용할 장치를 변수로 지정합니다.

``` python
import torch

# GPU가 사용 가능하면 'cuda', 아니면 'cpu'를 선택 (맥북 M1/M2/M3의 경우 'mps')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("사용 중인 장치:", device)
```

### 2). 데이터(Tensor)를 GPU로 이동
- 생성한 텐서에 .to(device)를 붙여 GPU 메모리로 보냅니다.

```python
x = torch.tensor([1.0, 2.0, 3.0])
x = x.to(device)  # x를 GPU로 이동

```

### 3). 모델을 GPU로 이동
- 마찬가지로 정의한 신경망 모델도 .to(device)로 GPU에 올려야 합니다.

``` python
model = MyModel()
model = model.to(device)  # 모델의 가중치들을 GPU로 이동
```

- 파이토치는 같은 장치에 있는 데이터끼리만 연산할 수 있기 때문에, [입력 데이터 + 모델 가중치]가 모두 동일한 GPU 메모리에 올라가 있는지 늘 확인하셔야 합니다.
- 결론적으로, "장치 지정(device) $\rightarrow$ 데이터와 모델을 .to(device)로 보내기" 이 패턴만 기억하고 계시면 됩니다!

````python
# 3. 텐서 생성법 
# 기본 함수 : 출력 텐서의 모양을 입력 받아 텐서를 반환
# _like 함수 : 다른 텐서를 입력 받아 입력 텐서와 같은 모양의 텐서를 반환
# (매번 텐서 크기를 지정하기 힘드니까 입력 텐서랑 같은 크기로 만듬.)

# empty 함수 : 임의의 값으로 채워져 있음
t= torch.empty([3,4])
print(t)
print(torch.empty_like(t))
print(torch.empty(t.shape))
````

````python
# 텐서 합치기
# cat
x= torch.randn(2,3)
print(x)
print(torch.cat((x,x,x), dim=0)) # 첫 인자 : 결합하고자 하는 텐서, 두번째 인자 : 어떤 차원인지.
print(torch.cat((x,x,x),0).shape)
````
