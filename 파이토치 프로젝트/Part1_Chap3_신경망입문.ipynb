## 1. 퍼셉트론의 기본 개념
### 0). 논리 게이트
- 논리 게이트를 퍼셉트론으로 구현 해본다.


```python
# AND 게이트
def AND_gate(x1,x2) :
    # weights
    w1=0.5
    w2=0.5
    bias = -0.6

    # forward
    result = x1 *w1 + x2 *w2 + bias
    # activation
    if result > 0 :
        return 1
    else : 
        return 0
print(AND_gate(0,0), AND_gate(0,1), AND_gate(1,0), AND_gate(1,1))
```

```python
# OR 게이트
def OR_gate(x1,x2):
    # weights
    w1=0.6
    w2=0.6
    bias = -0.5

    #forward
    result = x1*w1 + x2*w2 + bias
    #activation
    if result >0:
        return 1
    else :
        return 0

print(OR_gate(0,0), OR_gate(0,1), OR_gate(1,0), OR_gate(1,1))

```

```python
# NOT 게이트
def NOT_gate(x1):
    # weights
    w1=-0.6
    bias = 0.5

    #forward
    result = x1*w1  + bias
    #activation
    if result >0:
        return 1
    else :
        return 0

NOT_gate(0), NOT_gate(1)
```

## 2.XOR 문제
- 위 코드에서 본 것과 같은 선형으로 구분할 수 있는 문제는 퍼셉트론으로 해결할 수 있다.
- 이때, 선형은 서로 다른 분류를 구분하는 경계이며 이를 결정 경계라고 부른다.
- 이 결정 경계가 선형이라면 퍼셉트론으로 해결할 수 있으나, XOR 같은 문제는 풀지 못한다.
- 이를 해결하기 위해 다층 퍼셉트론을 사용한다.
- 총 2개의 Layer를 사용한다. 은닉층과 출력층. (입력 레이어는 세지 않는다.)
- 한 레이어당 2개의 퍼셉트론
- XOR 문제를 풀기 위해 NAND게이트와 OR게이트를 사용하면 된다.

```python
def Nand_gate(x1,x2):
    #weights
    w1=-0.5
    w2=-0.5
    b=0.6

    #forward
    result = x1*w1 + x2*w2 + b

    #activation
    if result > 0 :
        return 1
    else : 
        return 0
Nand_gate(0,0),Nand_gate(0,1),Nand_gate(1,0),Nand_gate(1,1)
```
